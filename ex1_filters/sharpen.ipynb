```python
def sharpen_kernel(n_size):

    kernel = np.array(np.zeros((n_size, n_size), dtype=np.float32))
    
    for i in range(n_size):
        for j in range(n_size):
            kernel[i][j] = 0 if i % 2 == 1 or j % 2 == 1 else -1

    kernel[n_size // 2, n_size // 2] = n_size * n_size - 1

    return kernel
```

```python
kernel = sharpen_kernel(5)
```

```python
sharpen_kernel = np.array([[0, -1, 0],
                           [-1, 5, -1],
                           [0, -1, 0]], dtype=np.float32)
```

```python
image_kernel = cv2.filter2D(base_image, -1, sharpen_kernel)

plt.imshow(image_kernel, cmap='gray')
```
