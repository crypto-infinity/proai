```python
import numpy as np
import gymnasium as gym
import matplotlib.pyplot as plt
```

```python
def policy_evaluation(V, env, policy, g, e):

    n_states = env.observation_space.n

    while True:
        delta = 0

        for state in range(n_states):
            v = V[state]
            action = policy[state]
            V[state] = 0 #errore

            for prob, next_state, reward, done in env.P[state][action]:
                V[state] += prob * (reward + g * V[next_state]) #errore

            delta = max(delta, abs(v - V[state]))

        if delta < e:
            break
    
    return V
```

```python
def policy_improvement(V, env, g, e):

    n_states = env.observation_space.n
    n_actions = env.action_space.n

    policy_stable = True
    policy = np.zeros(n_states, dtype=int)

    for state in range(n_states):

        old_action = policy[state]
        q_values = np.zeros(n_actions)

        for action in range(n_actions):
            for prob, next_state, reward, done in env.P[state][action]:
                q_values[action] += prob * (reward + g * V[next_state]) #q_values errore 

        best_action = np.argmax(q_values) #la value function
        policy[state] = best_action

        if old_action != best_action: 
            policy_stable = False

        policy[state] = best_action

    return policy_stable, policy
```

```python
def policy_iteration(env, g=0.99, e=1e-6):
    """
    """

    # 1 - initialization

    n_states = env.observation_space.n
    n_actions = env.action_space.n
    V = np.zeros(n_states)
    policy = np.zeros(n_states, dtype=int) #errore
    iteration = 0

    while True:

        V = policy_evaluation(V, env, policy, g, e) # 2 - Policy Evaluation

        policy_stable, policy = policy_improvement(V, env, g, e) # 3 - Policy improvement

        iteration += 1
        
        if policy_stable:
            break

    return V, policy, iteration
```
