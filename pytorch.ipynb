```python
cuda = torch.device('cuda')   

cuda
```

```python
a = torch.tensor([1., 2.])
print(a)

with cuda:
    # This will run on the GPU

    b = torch.tensor([1., 2.])
    print(b)

    c = torch.tensor([[1., 2.], [3., 4.]])
    print(c)

    d = torch.matmul(b, c)
    print(d)
```

```python
# set seed for reproducibility
torch.manual_seed(15)

a = torch.rand(5,5)

b = torch.rand(1,5)

print(torch.matmul(a, b.T))
```

```python
a = torch.rand(2,4)

b = torch.rand(2,4)

a = a.to(cuda)
b = b.to(cuda)

print(a, b)
```

```python
c = torch.mul(a, b)

print(torch.max(c), torch.min(c))
print(torch.mean(c))
```
